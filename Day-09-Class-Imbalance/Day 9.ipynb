```python
import csv

with open("train_fraud.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    train_data = list(reader)

print("Training records:", len(train_data))
print("Number of columns:", len(train_data[0]))
print("Target column:", "is_fraud")
```

```python
from collections import Counter

class_counts = Counter(row["is_fraud"] for row in train_data)

non_fraud_count = class_counts.get("0", 0)
fraud_count = class_counts.get("1", 0)

print("Training Class Distribution")
print("----------------------------")
print("Non-Fraud (0):", non_fraud_count)
print("Fraud (1):", fraud_count)

print("\nFraud Percentage:",
      round((fraud_count / len(train_data)) * 100, 2), "%")

print("Non-Fraud Percentage:",
      round((non_fraud_count / len(train_data)) * 100, 2), "%")
```

```python
# Separate the training data into two classes

non_fraud_data = [
    row for row in train_data
    if row["is_fraud"] == "0"
]

fraud_data = [
    row for row in train_data
    if row["is_fraud"] == "1"
]

print("Non-Fraud records:", len(non_fraud_data))
print("Fraud records:", len(fraud_data))
```

```python
# Target size for the minority class
target_fraud_count = len(non_fraud_data)

current_fraud_count = len(fraud_data)

additional_fraud_needed = target_fraud_count - current_fraud_count

print("Current fraud records:", current_fraud_count)
print("Target fraud records:", target_fraud_count)
print("Additional fraud records needed:", additional_fraud_needed)
```

```python
import random

# Make the result reproducible
random.seed(42)

# Randomly select additional fraud records with replacement
oversampled_fraud = random.choices(
    fraud_data,
    k=additional_fraud_needed
)

# Add the original fraud records
balanced_fraud_data = fraud_data + oversampled_fraud

print("Original fraud records:", len(fraud_data))
print("Additional fraud records:", len(oversampled_fraud))
print("Total fraud records after oversampling:", len(balanced_fraud_data))
```

```python
# Combine non-fraud and oversampled fraud records

balanced_train_data = non_fraud_data + balanced_fraud_data

# Shuffle the balanced dataset
random.seed(42)
random.shuffle(balanced_train_data)

print("Balanced training dataset created!")
print("Total records:", len(balanced_train_data))
```

```python
from collections import Counter

balanced_counts = Counter(
    row["is_fraud"] for row in balanced_train_data
)

non_fraud_balanced = balanced_counts.get("0", 0)
fraud_balanced = balanced_counts.get("1", 0)

print("Final Balanced Training Distribution")
print("-------------------------------------")
print("Non-Fraud (0):", non_fraud_balanced)
print("Fraud (1):", fraud_balanced)

print("\nNon-Fraud Percentage:",
      round((non_fraud_balanced / len(balanced_train_data)) * 100, 2), "%")

print("Fraud Percentage:",
      round((fraud_balanced / len(balanced_train_data)) * 100, 2), "%")
```

```python
import csv

with open("train_fraud_balanced.csv", "w", newline="", encoding="utf-8") as file:
    writer = csv.DictWriter(
        file,
        fieldnames=balanced_train_data[0].keys()
    )
    
    writer.writeheader()
    writer.writerows(balanced_train_data)

print("✅ Balanced training dataset saved successfully!")
print("File: train_fraud_balanced.csv")
print("Rows:", len(balanced_train_data))
```
